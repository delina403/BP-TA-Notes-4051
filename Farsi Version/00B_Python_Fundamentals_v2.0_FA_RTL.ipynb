## ‏بخش ۱ · پایتون و ابزارهای اطراف آن

### ‏۱.۵ · خواندن چند خط کد پایتون

- **عبارت (expression)** مقداری محاسبه می‌کند: <span dir="ltr" style="unicode-bidi: isolate;">`2 + 3 * 4`</span> برابر <span dir="ltr" style="unicode-bidi: isolate;">`14`</span> است.
- **دستور (statement)** عملی انجام می‌دهد: <span dir="ltr" style="unicode-bidi: isolate;">`print(14)`</span> عدد را نمایش می‌دهد؛ انتساب را نیز به‌زودی یاد می‌گیریم.
- متن بعد از <span dir="ltr" style="unicode-bidi: isolate;">`#`</span> یک **توضیح (comment)** است و پایتون آن را اجرا نمی‌کند.
- در پایتون **املای نام‌ها، حروف بزرگ/کوچک، گیومه، پرانتز** و (در ساختارهای بلوکی مثل حلقه‌ها) **تورفتگی** مهم‌اند. عرف رایج برای هر سطح تورفتگی **چهار فاصله** است.
- در یک فایل ساده، دستورهای سطح اصلی **به‌ترتیب از بالا به پایین** اجرا می‌شوند. **نکتهٔ ظریف:** پایتون پیش از اجرا کد را تحلیل/کامپایل می‌کند؛ بنابراین یک <span dir="ltr" style="unicode-bidi: isolate;">`SyntaxError`</span> در انتهای فایل ممکن است نگذارد خط‌های اول هم اجرا شوند. اما خطای زمان اجرا وقتی به آن عملیات برسیم اتفاق می‌افتد.

**پیش‌بینی کنید:** برنامهٔ بعدی چند خط خروجی تولید می‌کند؟

</div>


````python
# This is a comment, not output.
print("Start")
print(7 * (2 + 1))  # parentheses first
print("Finish")
````

## ‏بخش ۲ · متغیرها: نام‌گذاری و نگهداری مقدارها
### ‏۲.۱ · متغیر، نامی برای اشاره به یک مقدار است

در انتساب <span dir="ltr" style="unicode-bidi: isolate;">`distance = 12`</span>، ابتدا مقدار سمت راست محاسبه می‌شود و سپس نام <span dir="ltr" style="unicode-bidi: isolate;">`distance`</span> به آن مقدار **متصل** می‌شود. بهتر است متغیر را مانند یک **برچسب برای اشاره به یک شیء** تصور کنیم، نه جعبه‌ای با نوع ثابت.

- در پایتون لازم نیست نوع را مثل <span dir="ltr" style="unicode-bidi: isolate;">`int distance;`</span> از پیش اعلام کنیم.
- **مقدار/شیء نوع دارد**؛ یک نام بعداً می‌تواند به مقداری از نوع دیگر اشاره کند (هرچند تغییر پی‌درپی نوع خوانایی را کم می‌کند).
- <span dir="ltr" style="unicode-bidi: isolate;">`=`</span> یعنی **انتساب**، نه ادعای ریاضی «برابری دو طرف». برای *مقایسهٔ برابری* از <span dir="ltr" style="unicode-bidi: isolate;">`==`</span> استفاده می‌کنیم.
- استفاده از نامی که هنوز تعریف نشده باعث <span dir="ltr" style="unicode-bidi: isolate;">`NameError`</span> می‌شود.


<div dir="ltr" style="direction: ltr; text-align: left;">

```python
distance = 12
print(distance)  # مقدار را چاپ می‌کند، نه کلمهٔ distance
```

</div>

</div>


````python
distance = 12
name = "Ada"
print("distance:", distance)
print("name:", name)
print("type of distance:", type(distance))
````

### ‏۲.۲ · انتساب، انتساب دوباره و به‌روزرسانی

پایتون ابتدا **سمت راست** انتساب را محاسبه می‌کند و بعد نتیجه را به نامِ سمت چپ نسبت می‌دهد.


<div dir="ltr" style="direction: ltr; text-align: left;">

```python
counter = 4
counter = counter + 1  # ابتدا مقدار 4 خوانده می‌شود و سپس 5 جایگزین می‌شود
counter += 2           # کوتاه‌شدهٔ counter = counter + 2
```

</div>


در پایان، <span dir="ltr" style="unicode-bidi: isolate;">`counter`</span> برابر **7** است. عبارت <span dir="ltr" style="unicode-bidi: isolate;">`counter = counter + 1`</span> در برنامه‌نویسی معتبر است، اما **معادلهٔ ریاضی نیست**. دستور <span dir="ltr" style="unicode-bidi: isolate;">`+=`</span> نوعی *انتساب ترکیبی* است.

دو الگوی مفید: <span dir="ltr" style="unicode-bidi: isolate;">`a, b = 3, 4`</span> دو مقدار را انتساب می‌دهد و <span dir="ltr" style="unicode-bidi: isolate;">`a, b = b, a`</span> جای آن‌ها را عوض می‌کند. **نام را قبل از انتساب به کار نبرید**.

</div>


````python
counter = 4
counter = counter + 1
counter += 2
print("counter:", counter)
a, b = 3, 4
a, b = b, a
print("swapped:", a, b)
````

### ‏۲.۳ · نام متغیر: چه چیزی مجاز و چه چیزی خواناست؟

**قواعد زبان:** نام می‌تواند شامل حرف‌های یونیکد، رقم و زیرخط (<span dir="ltr" style="unicode-bidi: isolate;">`_`</span>) باشد، ولی **نباید با رقم آغاز شود**؛ فاصله، خط تیره و نمادهایی مثل <span dir="ltr" style="unicode-bidi: isolate;">`@`</span> مجاز نیستند. همچنین نام **نباید کلمهٔ کلیدی پایتون** مانند <span dir="ltr" style="unicode-bidi: isolate;">`for`</span>، <span dir="ltr" style="unicode-bidi: isolate;">`if`</span>، <span dir="ltr" style="unicode-bidi: isolate;">`class`</span>، <span dir="ltr" style="unicode-bidi: isolate;">`True`</span> یا <span dir="ltr" style="unicode-bidi: isolate;">`None`</span> باشد. حروف بزرگ و کوچک تفاوت دارند.

**توصیه‌های مهم برای کدنویسی خوانا** (نه الزام نحوی):

- نام‌های گویا با روش **<span dir="ltr" style="unicode-bidi: isolate;">`snake_case`</span>** بنویسید: <span dir="ltr" style="unicode-bidi: isolate;">`exam_score`</span> و <span dir="ltr" style="unicode-bidi: isolate;">`student_count`</span>.
- برای سازگاری و کار گروهی، نام‌های **انگلیسی با حروف ASCII** بهترند، هرچند پایتون بسیاری از حروف یونیکد را هم می‌پذیرد.
- نام‌های آمادهٔ پایتون مانند **<span dir="ltr" style="unicode-bidi: isolate;">`print`</span>, <span dir="ltr" style="unicode-bidi: isolate;">`list`</span>, <span dir="ltr" style="unicode-bidi: isolate;">`str`</span>, <span dir="ltr" style="unicode-bidi: isolate;">`input`</span>, <span dir="ltr" style="unicode-bidi: isolate;">`sum`</span>, <span dir="ltr" style="unicode-bidi: isolate;">`type`</span>** را متغیر نکنید. برخلاف کلمات کلیدی، این کار ممکن است اما عملکرد معمول آن‌ها را خراب می‌کند.
- نامی مثل <span dir="ltr" style="unicode-bidi: isolate;">`MAX_ATTEMPTS = 3`</span> طبق قرارداد **ثابت** در نظر گرفته می‌شود، ولی پایتون اجباری برای ثابت‌ماندن آن ندارد.

### ‏۲.۴ · انتساب با برابری فرق دارد؛ نام با متن یکی نیست

ببینید چه چیزی **تغییر می‌کند** و چه چیزی **نه**:

- <span dir="ltr" style="unicode-bidi: isolate;">`x = 8`</span> به یک نام مقدار نسبت می‌دهد؛ <span dir="ltr" style="unicode-bidi: isolate;">`x == 8`</span> **سؤال** می‌پرسد و <span dir="ltr" style="unicode-bidi: isolate;">`True`</span> یا <span dir="ltr" style="unicode-bidi: isolate;">`False`</span> نتیجه می‌دهد.
- <span dir="ltr" style="unicode-bidi: isolate;">`y = x`</span> مقداری را که اکنون <span dir="ltr" style="unicode-bidi: isolate;">`x`</span> به آن اشاره می‌کند به <span dir="ltr" style="unicode-bidi: isolate;">`y`</span> نیز نسبت می‌دهد؛ برای عدد صحیح، انتساب دوباره به <span dir="ltr" style="unicode-bidi: isolate;">`x`</span> در ادامه **<span dir="ltr" style="unicode-bidi: isolate;">`y`</span> را تغییر نمی‌دهد**.
- <span dir="ltr" style="unicode-bidi: isolate;">`print(x)`</span> مقدار را چاپ می‌کند؛ <span dir="ltr" style="unicode-bidi: isolate;">`print("x")`</span> حرف <span dir="ltr" style="unicode-bidi: isolate;">`x`</span> را به‌صورت متن نمایش می‌دهد.
- <span dir="ltr" style="unicode-bidi: isolate;">`print(z)`</span> قبل از تعریف <span dir="ltr" style="unicode-bidi: isolate;">`z`</span> باعث **<span dir="ltr" style="unicode-bidi: isolate;">`NameError`</span>** می‌شود؛ پایتون نمی‌تواند حدس بزند <span dir="ltr" style="unicode-bidi: isolate;">`z`</span> چیست.

> وقتی بعداً شیءهای *تغییرپذیر* مثل فهرست را یاد بگیریم، بحث اشاره‌کردن چند نام به یک شیء را دقیق‌تر بررسی می‌کنیم. فعلاً: **انتساب یعنی اتصال نام به مقدار، نه تساوی جبری.**

</div>


````python
x = 8
y = x
print("Equality test:", x == 8)
x = 10
print("x:", x, "y:", y)
print("x")          # literal text
print(x)            # the value now bound to x
````

### ‏۳.۳ · نکتهٔ مخصوص دانشجویان ریاضی: صحیحِ دقیق و شناورِ تقریبی

- **اعداد صحیح** را (تا حد حافظهٔ موجود) می‌توان به‌صورت دقیق نمایش داد.
- بیشتر **کسرهای ده‌دهی** در نمایش دودوییِ <span dir="ltr" style="unicode-bidi: isolate;">`float`</span> کاملاً دقیق نیستند. بنابراین <span dir="ltr" style="unicode-bidi: isolate;">`0.1 + 0.2`</span> در داخل رایانه اندکی با <span dir="ltr" style="unicode-bidi: isolate;">`0.3`</span> فرق دارد.
- برای مقایسهٔ عددیِ تقریبی، معمولاً <span dir="ltr" style="unicode-bidi: isolate;">`math.isclose(a, b)`</span> با آستانهٔ متناسب با مسئله بهتر است. هنگام مقایسهٔ اعداد نزدیک صفر، **تلورانس مطلق** اهمیت ویژه دارد.
- <span dir="ltr" style="unicode-bidi: isolate;">`round(x, n)`</span> و <span dir="ltr" style="unicode-bidi: isolate;">`f"{x:.2f}"`</span> مشکل دقت داخلی را برطرف نمی‌کنند؛ دومی فقط **نمایش خروجی** را کنترل می‌کند.

**پیش‌بینی کنید:** آیا <span dir="ltr" style="unicode-bidi: isolate;">`0.1 + 0.2 == 0.3`</span> برابر <span dir="ltr" style="unicode-bidi: isolate;">`True`</span> می‌شود یا <span dir="ltr" style="unicode-bidi: isolate;">`False`</span>؟ سپس مثال را اجرا کنید.

</div>


````python
import math  # Python's standard math library

print("0.1 + 0.2 =", 0.1 + 0.2)
print("Exact comparison:", 0.1 + 0.2 == 0.3)
print("Approximate comparison:", math.isclose(0.1 + 0.2, 0.3))
print("Very large integer:", 10 ** 25)
print("Formatted result:", f"{0.1 + 0.2:.2f}")
````

### ‏۳.۴ · رشته‌ها: متن با عدد فرق دارد

رشته (<span dir="ltr" style="unicode-bidi: isolate;">`str`</span>) دنباله‌ای از **نویسه‌های یونیکد** است. آن را در گیومهٔ تکی یا دوتایی می‌نویسیم؛ سه گیومهٔ هماهنگ هم برای متن چندخطی کاربرد دارد. چند عملیات مفید:

- <span dir="ltr" style="unicode-bidi: isolate;">`"py" + "thon"` → `"python"`</span> (**چسباندن دو رشته**).
- <span dir="ltr" style="unicode-bidi: isolate;">`"ha" * 3` → `"hahaha"`</span> (**تکرار**).
- <span dir="ltr" style="unicode-bidi: isolate;">`len("cat")` → `3`</span> (**طول**؛ در این مثال سه کدپوینت یونیکد).
- <span dir="ltr" style="unicode-bidi: isolate;">`"cat"[0]` → `"c"`</span> (شماره‌گذاری از **صفر** شروع می‌شود)؛ <span dir="ltr" style="unicode-bidi: isolate;">`"cat"[-1]` → `"t"`</span>.
- <span dir="ltr" style="unicode-bidi: isolate;">`"python"[1:4]` → `"yth"`</span> (ابتدا **داخل** بازه، انتها **خارج** بازه است).
- توالی‌های ویژه: <span dir="ltr" style="unicode-bidi: isolate;">`\n`</span> رفتن به خط جدید، <span dir="ltr" style="unicode-bidi: isolate;">`\t`</span> تب، <span dir="ltr" style="unicode-bidi: isolate;">`\\`</span> بک‌اسلش و <span dir="ltr" style="unicode-bidi: isolate;">`\"`</span> برای قراردادن گیومه در متن.

**دام‌های رایج:** <span dir="ltr" style="unicode-bidi: isolate;">`"5" + "2"`</span> به <span dir="ltr" style="unicode-bidi: isolate;">`"52"`</span> می‌رسد، نه <span dir="ltr" style="unicode-bidi: isolate;">`7`</span>. دستور <span dir="ltr" style="unicode-bidi: isolate;">`"5" + 2`</span> خطای <span dir="ltr" style="unicode-bidi: isolate;">`TypeError`</span> دارد. اندیس ناموجود مثل <span dir="ltr" style="unicode-bidi: isolate;">`"cat"[3]`</span> خطای <span dir="ltr" style="unicode-bidi: isolate;">`IndexError`</span> می‌دهد. رشته **تغییرناپذیر** است و <span dir="ltr" style="unicode-bidi: isolate;">`word[0] = "C"`</span> مجاز نیست.

</div>


````python
word = "Python"
print("Concatenation:", "py" + "thon")
print("Repetition:", "ha" * 3)
print("Length:", len(word))
print("First and last:", word[0], word[-1])
print("Slice [1:4]:", word[1:4])
print("Text, not arithmetic:", "5" + "2")
print("Two lines:\nsecond line")
````

### ‏۳.۵ · بولی، مقایسه و عملگرهای منطقی

نوع بولی فقط دو مقدار دارد: **<span dir="ltr" style="unicode-bidi: isolate;">`True`</span>** و **<span dir="ltr" style="unicode-bidi: isolate;">`False`</span>** (نه <span dir="ltr" style="unicode-bidi: isolate;">`true`</span> و <span dir="ltr" style="unicode-bidi: isolate;">`false`</span> و نه رشته‌های <span dir="ltr" style="unicode-bidi: isolate;">`"True"`</span> و <span dir="ltr" style="unicode-bidi: isolate;">`"False"`</span>). نتیجهٔ مقایسه‌ها بولی است:

<span dir="ltr" style="unicode-bidi: isolate;">`==`</span> مساوی · <span dir="ltr" style="unicode-bidi: isolate;">`!=`</span> نامساوی · <span dir="ltr" style="unicode-bidi: isolate;">`<`</span> کوچک‌تر · <span dir="ltr" style="unicode-bidi: isolate;">`<=`</span> کوچک‌تر یا مساوی · <span dir="ltr" style="unicode-bidi: isolate;">`>`</span> بزرگ‌تر · <span dir="ltr" style="unicode-bidi: isolate;">`>=`</span> بزرگ‌تر یا مساوی.

منطق بولی: <span dir="ltr" style="unicode-bidi: isolate;">`and`</span> نیازمند درست‌بودن هر دو شرط است؛ <span dir="ltr" style="unicode-bidi: isolate;">`or`</span> با درست‌بودن حداقل یکی درست می‌شود؛ <span dir="ltr" style="unicode-bidi: isolate;">`not`</span> ارزش منطقی را وارونه می‌کند. می‌توان مثل ریاضی نوشت: <span dir="ltr" style="unicode-bidi: isolate;">`0 <= score <= 20`</span>.

**تبدیل به ارزش منطقی:** <span dir="ltr" style="unicode-bidi: isolate;">`bool(0)`</span> و <span dir="ltr" style="unicode-bidi: isolate;">`bool("")`</span> برابر <span dir="ltr" style="unicode-bidi: isolate;">`False`</span> هستند؛ عدد غیرصفر و رشتهٔ غیرخالی معمولاً <span dir="ltr" style="unicode-bidi: isolate;">`True`</span> هستند. **مراقب باشید:** <span dir="ltr" style="unicode-bidi: isolate;">`bool("False")`</span> برابر <span dir="ltr" style="unicode-bidi: isolate;">`True`</span> است، چون ورودی یک **رشتهٔ غیرخالی** است، نه مقدار بولی <span dir="ltr" style="unicode-bidi: isolate;">`False`</span>.

اگر عملوندها بولی نباشند، <span dir="ltr" style="unicode-bidi: isolate;">`and`</span> و <span dir="ltr" style="unicode-bidi: isolate;">`or`</span> می‌توانند خود **یکی از عملوندها** را برگردانند، نه الزاماً <span dir="ltr" style="unicode-bidi: isolate;">`bool`</span>. در این جلسه برای ترکیب **مقایسه‌ها** از آن‌ها استفاده کنید.

**نکتهٔ تکمیلی، غیرضروری برای این جلسه:** عملگر <span dir="ltr" style="unicode-bidi: isolate;">`is`</span> هویت شیء را مقایسه می‌کند؛ برای برابری عادیِ مقدارها از <span dir="ltr" style="unicode-bidi: isolate;">`==`</span> استفاده کنید. در پایتون <span dir="ltr" style="unicode-bidi: isolate;">`bool`</span> از نظر فنی زیرنوع <span dir="ltr" style="unicode-bidi: isolate;">`int`</span> است، ولی بولی را برای منطق به کار ببرید نه محاسبات عددی.

</div>


````python
score = 17
print("Pass (>= 10):", score >= 10)
print("In 0..20:", 0 <= score <= 20)
print("Both:", score >= 10 and score <= 20)
print("At least one:", score < 10 or score > 15)
print("Not equal:", score != 20)
print("String vs number:", "3" == 3)
print("bool(0):", bool(0))
print('bool("False"):', bool("False"))
````

### ‏۴.۱ · خروجی: تابع <span dir="ltr" style="unicode-bidi: isolate;">`print()`</span> اطلاعات را نمایش می‌دهد

تابع <span dir="ltr" style="unicode-bidi: isolate;">`print(...)`</span> آرگومان‌ها را محاسبه می‌کند، به شکل متن قابل‌نمایش درمی‌آورد و در خروجی استاندارد (معمولاً **ترمینال**) می‌نویسد.

- <span dir="ltr" style="unicode-bidi: isolate;">`print("Hello")`</span> متن را **بدون نمایش گیومه‌ها** چاپ می‌کند.
- <span dir="ltr" style="unicode-bidi: isolate;">`print("x =", 5)`</span> می‌تواند متن و عدد را با هم چاپ کند؛ بین آرگومان‌ها معمولاً **فاصله** قرار می‌گیرد.
- هر <span dir="ltr" style="unicode-bidi: isolate;">`print()`</span> به‌طور پیش‌فرض به **خط بعد** می‌رود.
- گزینهٔ <span dir="ltr" style="unicode-bidi: isolate;">`sep=`</span> جداکنندهٔ آرگومان‌ها را و <span dir="ltr" style="unicode-bidi: isolate;">`end=`</span> متن انتهای خروجی را تعیین می‌کند.
- <span dir="ltr" style="unicode-bidi: isolate;">`print()`</span> بدون آرگومان یک خط خالی ایجاد می‌کند.

> **نمایش با ذخیره‌کردن فرق دارد:** <span dir="ltr" style="unicode-bidi: isolate;">`print(5)`</span> عدد <span dir="ltr" style="unicode-bidi: isolate;">`5`</span> را نشان می‌دهد، ولی در متغیری ذخیره نمی‌کند. مقدار بازگشتیِ <span dir="ltr" style="unicode-bidi: isolate;">`print()`</span> برابر <span dir="ltr" style="unicode-bidi: isolate;">`None`</span> است.

</div>


````python
print("A", "B", 3)                  # default separator: a space
print("A", "B", 3, sep=" | ")
print("Hi", end="!")                 # no newline here
print(" Again")                      # continues on the same line
print()                              # a blank line
print("Done")
````

### ‏۴.۲ · خروجی خواناتر با **<span dir="ltr" style="unicode-bidi: isolate;">f-string</span>**

در **<span dir="ltr" style="unicode-bidi: isolate;">f-string</span>** می‌توان مقدار یک عبارت را داخل <span dir="ltr" style="unicode-bidi: isolate;">`{...}`</span> قرار داد و همراه توضیح چاپ کرد:


<div dir="ltr" style="direction: ltr; text-align: left;">

```python
name = "Ada"
score = 18.375
print(f"{name} scored {score:.2f} points")
```

</div>


خروجی: <span dir="ltr" style="unicode-bidi: isolate;">`Ada scored 18.38 points`</span> (نمایش با دو رقم اعشار). عبارت <span dir="ltr" style="unicode-bidi: isolate;">`:.2f`</span> فقط **شیوهٔ نمایش عدد شناور** را عوض می‌کند، نه مقدار ذخیره‌شده در <span dir="ltr" style="unicode-bidi: isolate;">`score`</span> را.

- <span dir="ltr" style="unicode-bidi: isolate;">`print("Score:", score)`</span> هم کاملاً درست است.
- <span dir="ltr" style="unicode-bidi: isolate;">`"Score: " + str(score)`</span> کار می‌کند، اما <span dir="ltr" style="unicode-bidi: isolate;">`"Score: " + score`</span> **کار نمی‌کند**.
- وجود حرف <span dir="ltr" style="unicode-bidi: isolate;">`f`</span> **قبل از گیومه** ضروری است؛ بدون آن، آکولادها متن عادی‌اند.

> خروجی باید برای انسان واضح باشد: برچسب و واحد کمیت‌ها را بنویسید و در صورت نیاز تعداد ارقام اعشار را تعیین کنید.

</div>


````python
name = "Ada"
score = 18.375
print(f"{name} scored {score:.2f} points")
print("Raw score:", score)
print(f"Squared score: {score ** 2:.1f}")
print("Without f:", "{name} scored {score:.2f} points")
````

### ‏۴.۳ · ورودی: تابع <span dir="ltr" style="unicode-bidi: isolate;">`input()`</span> **همیشه رشته** برمی‌گرداند

تابع <span dir="ltr" style="unicode-bidi: isolate;">`input(prompt)`</span> ابتدا پرسش را نمایش می‌دهد، **منتظر** می‌ماند تا کاربر چیزی بنویسد و Enter بزند، سپس همان متن را از نوع **<span dir="ltr" style="unicode-bidi: isolate;">`str`</span>** برمی‌گرداند. نویسهٔ پایان خط حذف می‌شود، ولی فاصله‌های اضافی ابتدا و انتهای متن خودکار حذف نمی‌شوند.


- اگر کاربر فقط Enter بزند، نتیجه <span dir="ltr" style="unicode-bidi: isolate;">`""`</span> (**رشتهٔ خالی**) است.
- <span dir="ltr" style="unicode-bidi: isolate;">`input()`</span> از روی ظاهر متن نمی‌فهمد که آیا ورودی عدد است یا نه.
- برای حذف فاصله‌های اطراف می‌توان از <span dir="ltr" style="unicode-bidi: isolate;">`.strip()`</span> استفاده کرد: <span dir="ltr" style="unicode-bidi: isolate;">`name = input("Name: ").strip()`</span>.
- تمرین را با **اجرای فایل <span dir="ltr" style="unicode-bidi: isolate;">`.py`</span> در ترمینال** انجام دهید، نه با اجرای همهٔ سلول‌های این دفترچه. کدهای تعاملی عمداً فقط در بخش‌های متنی آمده‌اند تا نمایشِ پروژکتوری متوقف نشود.

### ‏۴.۴ · ورودی عددی: **بعد از دریافت متن** تبدیل نوع کنید

اگر محاسبه لازم است، متن ورودی را صریحاً تبدیل کنید:


<div dir="ltr" style="direction: ltr; text-align: left;">

```python
age_text = input("Age: ")  # همیشه str
age = int(age_text)          # ممکن است ValueError بدهد
print(f"Next year: {age + 1}")
```

</div>


برای اندازه‌گیری‌هایی که ممکن است اعشاری باشند، از <span dir="ltr" style="unicode-bidi: isolate;">`float`</span> استفاده کنید:


<div dir="ltr" style="direction: ltr; text-align: left;">

```python
radius = float(input("Radius: "))
print(f"Radius squared: {radius ** 2:.2f}")
```

</div>


**قواعد و دام‌ها:**
- <span dir="ltr" style="unicode-bidi: isolate;">`int("  -12 ")`</span> معتبر است؛ <span dir="ltr" style="unicode-bidi: isolate;">`int("12.5")`</span> **نیست**. ولی <span dir="ltr" style="unicode-bidi: isolate;">`float("12.5")`</span> معتبر است.
- برای اعشار از **نقطه** (<span dir="ltr" style="unicode-bidi: isolate;">`3.5`</span>) استفاده کنید، نه ویرگول (<span dir="ltr" style="unicode-bidi: isolate;">`3,5`</span>).
- متن عددیِ نامعتبر مثل <span dir="ltr" style="unicode-bidi: isolate;">`"twelve"`</span> یا ورودی خالی هنگام تبدیل باعث <span dir="ltr" style="unicode-bidi: isolate;">`ValueError`</span> می‌شود.
- برای پرسش بله/خیر **از** <span dir="ltr" style="unicode-bidi: isolate;">`bool(input("yes/no?"))`</span> **استفاده نکنید**؛ تقریباً هر پاسخ غیرخالی حتی <span dir="ltr" style="unicode-bidi: isolate;">`"no"`</span> به <span dir="ltr" style="unicode-bidi: isolate;">`True`</span> تبدیل می‌شود.
- یک راه ساده: <span dir="ltr" style="unicode-bidi: isolate;">`answer = input("yes/no: ").strip().lower()`</span> و سپس <span dir="ltr" style="unicode-bidi: isolate;">`is_yes = (answer == "yes")`</span>.

> **فرض برنامه‌های اولیه:** کاربر مطابق راهنما عدد معتبر وارد می‌کند. تکرار پرسش، اعتبارسنجی و مدیریت استثناها را بعداً می‌آموزیم.

### ‏۴.۵ · برنامهٔ کاملِ گام‌به‌گام: هندسهٔ مستطیل

**مسئله:** با داشتن طول و عرض یک مستطیل (سانتی‌متر)، مساحت (سانتی‌متر مربع) و محیط (سانتی‌متر) را محاسبه کنید. کد را در فایل **<span dir="ltr" style="unicode-bidi: isolate;">`rectangle.py`</span>** ذخیره و در ترمینال اجرا کنید.

## ‏بخش ۵ · تمرین‌های مستقل با فایل <span dir="ltr" style="unicode-bidi: isolate;">`.py`</span>

برای **هر** تمرین: فایل جداگانه‌ای در VS Code بسازید، خروجی را **حدس بزنید**، با دست‌کم **دو ورودی آزمایشی** اجرا کنید و در یک جمله علت نتیجه را بنویسید. **فرمان شل را داخل فایل پایتون نگذارید.**

### ‏تمرین A · سن و خوشامدگویی (<span dir="ltr" style="unicode-bidi: isolate;">`age.py`</span>)

**نام** و **سنِ صحیح** کاربر را بگیرید؛ سلام کنید و سن او را ۵ سال بعد بنویسید. اگر <span dir="ltr" style="unicode-bidi: isolate;">`Ada`</span> و <span dir="ltr" style="unicode-bidi: isolate;">`20`</span> وارد شد، خط آخر باید <span dir="ltr" style="unicode-bidi: isolate;">`In 5 years, you will be 25.`</span> باشد. از ورودی متنی، تبدیل <span dir="ltr" style="unicode-bidi: isolate;">`int`</span> و <span dir="ltr" style="unicode-bidi: isolate;">`f-string`</span> استفاده کنید.

### ‏تمرین B · تبدیل دقیقه (<span dir="ltr" style="unicode-bidi: isolate;">`minutes.py`</span>)

یک **عدد صحیح نامنفی** بر حسب دقیقه بگیرید. با <span dir="ltr" style="unicode-bidi: isolate;">`//`</span> و <span dir="ltr" style="unicode-bidi: isolate;">`%`</span> ساعت کامل و دقیقهٔ باقیمانده را چاپ کنید. نمونه: <span dir="ltr" style="unicode-bidi: isolate;">`135` → `2 hours and 15 minutes`</span>. ورودی‌های <span dir="ltr" style="unicode-bidi: isolate;">`0`</span>، <span dir="ltr" style="unicode-bidi: isolate;">`59`</span>، <span dir="ltr" style="unicode-bidi: isolate;">`60`</span> و <span dir="ltr" style="unicode-bidi: isolate;">`135`</span> را آزمایش کنید.

### ‏تمرین C · میانگین وزنی (<span dir="ltr" style="unicode-bidi: isolate;">`weighted_mean.py`</span>)

دو نمرهٔ امتحان را از نوع <span dir="ltr" style="unicode-bidi: isolate;">`float`</span> بخوانید. وزن نمرهٔ اول **۳** و نمرهٔ دوم **۲** است. مقدار <span dir="ltr" style="unicode-bidi: isolate;">`(3*a + 2*b) / 5`</span> را محاسبه و با **دو رقم اعشار** چاپ کنید. نمونه: <span dir="ltr" style="unicode-bidi: isolate;">`a = 12`</span> و <span dir="ltr" style="unicode-bidi: isolate;">`b = 18` → `14.40`</span>. فکر کنید چرا مخرج ۵ است.

### ‏تمرین D · فاصلهٔ اقلیدسی (<span dir="ltr" style="unicode-bidi: isolate;">`distance.py`</span>) — چالش

چهار عدد شناور <span dir="ltr" style="unicode-bidi: isolate;">`x1`</span>، <span dir="ltr" style="unicode-bidi: isolate;">`y1`</span>، <span dir="ltr" style="unicode-bidi: isolate;">`x2`</span> و <span dir="ltr" style="unicode-bidi: isolate;">`y2`</span> را به‌عنوان مختصات دو نقطه در صفحه بگیرید. فاصلهٔ آن‌ها را با دو رقم اعشار و فقط با ابزارهای امروز محاسبه کنید:

<span dir="ltr" style="unicode-bidi: isolate;">`distance = ((x2 - x1) ** 2 + (y2 - y1) ** 2) ** 0.5`</span>

آزمایش کنید: <span dir="ltr" style="unicode-bidi: isolate;">`(0, 0)`</span> تا <span dir="ltr" style="unicode-bidi: isolate;">`(3, 4)` → `5.00`</span> و دو نقطهٔ یکسان ← <span dir="ltr" style="unicode-bidi: isolate;">`0.00`</span>. اگر زمان داشتید، مجذور فاصله را هم چاپ کنید.

## ‏جمع‌بندی · مدل ذهنی جلسهٔ 00B

<span dir="rtl" style="unicode-bidi: isolate;">`کد منبع (.py) ← محیط اجرای پایتون ← اجرای دستورها ← مقدارها و نوع‌ها ← خروجی`</span>

<span dir="ltr" style="unicode-bidi: isolate;">`input()`</span> **همیشه متن می‌دهد** ← در صورت نیاز با <span dir="ltr" style="unicode-bidi: isolate;">`int()`</span> یا <span dir="ltr" style="unicode-bidi: isolate;">`float()`</span> تبدیل می‌کنیم ← **محاسبه** ← نتیجه را با <span dir="ltr" style="unicode-bidi: isolate;">`print()`</span> یا **<span dir="ltr" style="unicode-bidi: isolate;">f-string</span>** نشان می‌دهیم.

**برای جلسهٔ بعد آماده‌اید اگر بتوانید:** تفاوت کامپایلر/مفسر را توضیح دهید و با ویرایشگر/ترمینال اشتباه نگیرید؛ متغیر خوانا بسازید؛ <span dir="ltr" style="unicode-bidi: isolate;">`int`</span>، <span dir="ltr" style="unicode-bidi: isolate;">`float`</span>، <span dir="ltr" style="unicode-bidi: isolate;">`str`</span> و <span dir="ltr" style="unicode-bidi: isolate;">`bool`</span> را تشخیص دهید؛ <span dir="ltr" style="unicode-bidi: isolate;">`/`</span>، <span dir="ltr" style="unicode-bidi: isolate;">`//`</span>، <span dir="ltr" style="unicode-bidi: isolate;">`%`</span> و <span dir="ltr" style="unicode-bidi: isolate;">`**`</span> را محاسبه کنید؛ انتساب‌ها را مرحله‌به‌مرحله دنبال کنید؛ ورودی معتبر را تبدیل کنید؛ پیام خطاهای متداول را بخوانید؛ و مستقلاً یک فایل <span dir="ltr" style="unicode-bidi: isolate;">`.py`</span> ساده بسازید و اجرا کنید.

> **روش مطالعه:** پیش از اجرای هر مثال تازه، روی کاغذ (۱) نوع هر مقدار، (۲) مقدار هر متغیر پس از هر خط و (۳) خروجی دقیق را بنویسید. سپس اجرا کنید و تفاوتِ پیش‌بینی و نتیجه را برای خود توضیح دهید.
