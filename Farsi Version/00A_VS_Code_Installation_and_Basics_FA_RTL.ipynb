# 🧭 جلسهٔ صفر A | نصب و کار با VS Code برای Python

**درس:** برنامه‌نویسی مقدماتی · **سطح:** بدون هیچ سابقهٔ برنامه‌نویسی · **زمان مطالعه و تمرین:** ۹۰ تا ۱۵۰ دقیقه

**این دفتر را از بالا به پایین بخوانید.** اگر هنوز VS Code یا Python نصب نکرده‌اید، بخش‌های ۱ تا ۴ را ابتدا در راهنمای مرورگری <span dir="ltr" style="unicode-bidi: isolate;">`START_HERE_VS_CODE.html`</span> یا فایل <span dir="ltr" style="unicode-bidi: isolate;">`START_HERE_VS_CODE.md`</span> بخوانید (این فایل‌ها بدون نصب Python/Jupyter باز می‌شوند).

### 🎯 بعد از این جلسه می‌توانید

- تفاوت **<span dir="ltr" style="unicode-bidi: isolate;">Python</span>**، **<span dir="ltr" style="unicode-bidi: isolate;">VS Code</span>**، **<span dir="ltr" style="unicode-bidi: isolate;">Python extension</span>** و **<span dir="ltr" style="unicode-bidi: isolate;">Jupyter kernel</span>** را توضیح دهید.
- VS Code و Python 3 را روی **<span dir="ltr" style="unicode-bidi: isolate;">Windows / macOS / Linux</span>** نصب و نصب را بررسی کنید.
- یک پوشهٔ پروژه بسازید؛ فایل <span dir="ltr" style="unicode-bidi: isolate;">`.py`</span> را ذخیره و از ترمینال اجرا کنید.
- دفتر <span dir="ltr" style="unicode-bidi: isolate;">`.ipynb`</span> را باز کنید، **<span dir="ltr" style="unicode-bidi: isolate;">Kernel</span>** مناسب انتخاب کنید و سلول‌ها را اجرا کنید.
- خطاهای متداول را بخوانید، **<span dir="ltr" style="unicode-bidi: isolate;">Interpreter</span>** را پیدا کنید و در صورت نیاز اشکال‌زدایی کنید.
- چند تمرین با بررسی خودکار حل کنید و یک برنامهٔ کوچک را در VS Code اجرا کنید.

> مسیر پیشنهادی: **نصب ← اجرای <span dir="ltr" style="unicode-bidi: isolate;">`hello.py`</span> ← اجرای همین دفتر ← تمرین ← بررسی چک‌لیست**. اگر VS Code را قبلاً نصب کرده‌اید، از بخش ۵ شروع کنید.

**نکتهٔ مهم:** برای استفاده از امکانات اصلی VS Code نه حساب کاربری لازم است و نه ابزارهای هوش مصنوعی. هر وقت با عبارت <span dir="ltr" style="unicode-bidi: isolate;">`Terminal`</span> یا <span dir="ltr" style="unicode-bidi: isolate;">`Command Palette`</span> روبه‌رو شدید، منظور بخشی از خود VS Code است.

## ۲. نصب: یک سیستم‌عامل را انتخاب کنید

### <span dir="ltr" style="unicode-bidi: isolate;">🪟 Windows 10/11</span>

1. به [صفحهٔ دریافت Python](https://www.python.org/downloads/) بروید و روش نصب پیشنهادی ویندوز را دنبال کنید. در نسخه‌های جدید، **<span dir="ltr" style="unicode-bidi: isolate;">Python install manager</span>** ممکن است ابتدا نصب شود و سپس نسخهٔ Python را نصب/مدیریت کند؛ اگر برنامه از شما خواست Python 3 را نصب کنید، مراحل را ادامه دهید. **نیازی نیست از روی راهنمای قدیمی به‌زور PATH را تغییر دهید.**
2. **<span dir="ltr" style="unicode-bidi: isolate;">PowerShell</span>** یا **<span dir="ltr" style="unicode-bidi: isolate;">Command Prompt</span>** را باز کنید و یکی از فرمان‌های زیر را امتحان کنید:


   <div dir="ltr" style="direction: ltr; text-align: left;">

   ```powershell
   py --version
   python --version
   ```

   </div>


   یک خروجی شبیه <span dir="ltr" style="unicode-bidi: isolate;">`Python 3.x.y`</span> نشانهٔ موفقیت است. در بعضی سیستم‌ها فقط یکی از این دستورها در دسترس است.
3. <span dir="ltr" style="unicode-bidi: isolate;">[VS Code Stable](https://code.visualstudio.com/download)</span> را دریافت کنید؛ **<span dir="ltr" style="unicode-bidi: isolate;">User Installer</span>** مناسب معماری رایانه را انتخاب و نصب کنید. سپس VS Code را از Start باز کنید.
4. در VS Code: آیکون **<span dir="ltr" style="unicode-bidi: isolate;">Extensions</span>** (چهار مربع) ← جست‌وجوی <span dir="ltr" style="unicode-bidi: isolate;">`Python`</span> ← افزونهٔ **<span dir="ltr" style="unicode-bidi: isolate;">Python</span>** با ناشر **<span dir="ltr" style="unicode-bidi: isolate;">Microsoft</span>** ← **<span dir="ltr" style="unicode-bidi: isolate;">Install</span>**. همین کار را برای **<span dir="ltr" style="unicode-bidi: isolate;">Jupyter</span>** با ناشر **<span dir="ltr" style="unicode-bidi: isolate;">Microsoft</span>** انجام دهید.

**خطای رایج:** فرمان <span dir="ltr" style="unicode-bidi: isolate;">`python`</span> به Microsoft Store می‌رود یا پیدا نمی‌شود. قبل از نصب دوباره، <span dir="ltr" style="unicode-bidi: isolate;">`py --version`</span> را امتحان کنید و راهنمای رسمی عیب‌یابی Python ویندوز را بخوانید: <span dir="ltr" style="unicode-bidi: isolate;">[Python on Windows](https://docs.python.org/3/using/windows.html)</span>. اگر Python را تازه نصب کردید، ترمینال جدید باز کنید.

</div>


<div dir="rtl" lang="fa" style="direction: rtl; text-align: right;">

### <span dir="ltr" style="unicode-bidi: isolate;">🍎 macOS (Intel / Apple silicon)</span>

1. Python 3 را از <span dir="ltr" style="unicode-bidi: isolate;">[python.org/downloads](https://www.python.org/downloads/)</span> دریافت کنید و بستهٔ macOS را طبق راهنمای نصب اجرا کنید. ممکن است macOS از قبل دستوری به نام <span dir="ltr" style="unicode-bidi: isolate;">`python3`</span> داشته باشد؛ **نسخه و مسیر** را بررسی کنید تا مطمئن شوید از نسخهٔ مناسب استفاده می‌کنید.
2. برنامهٔ VS Code را از <span dir="ltr" style="unicode-bidi: isolate;">[code.visualstudio.com/download](https://code.visualstudio.com/download)</span> بگیرید؛ فایل <span dir="ltr" style="unicode-bidi: isolate;">`.dmg`</span> را باز کنید و **<span dir="ltr" style="unicode-bidi: isolate;">Visual Studio Code.app</span>** را به **<span dir="ltr" style="unicode-bidi: isolate;">Applications</span>** بکشید.
3. برنامهٔ **<span dir="ltr" style="unicode-bidi: isolate;">Terminal</span>** را باز کنید و بزنید:


   <div dir="ltr" style="direction: ltr; text-align: left;">

   ```bash
   python3 --version
   ```

   </div>


4. در VS Code دو افزونهٔ **<span dir="ltr" style="unicode-bidi: isolate;">Python</span>** و **<span dir="ltr" style="unicode-bidi: isolate;">Jupyter</span>** (ناشر Microsoft) را نصب کنید.

**نکته:** در macOS معمولاً <span dir="ltr" style="unicode-bidi: isolate;">`python3`</span> مطمئن‌تر از <span dir="ltr" style="unicode-bidi: isolate;">`python`</span> است. اگر از محیط مجازی استفاده می‌کنید، دستور داخل محیط ممکن است <span dir="ltr" style="unicode-bidi: isolate;">`python`</span> باشد.

راهنما: <span dir="ltr" style="unicode-bidi: isolate;">[VS Code installation](https://code.visualstudio.com/docs/getstarted/overview)</span>

</div>


<div dir="rtl" lang="fa" style="direction: rtl; text-align: right;">

### 🐧 Linux (مثلاً Ubuntu/Debian یا Fedora)

1. با بسته‌های سیستم‌عامل خود **<span dir="ltr" style="unicode-bidi: isolate;">Python 3</span>**، ابزار ساخت محیط مجازی و <span dir="ltr" style="unicode-bidi: isolate;">`pip`</span> را آماده کنید. برای **<span dir="ltr" style="unicode-bidi: isolate;">Ubuntu/Debian</span>**:


   <div dir="ltr" style="direction: ltr; text-align: left;">

   ```bash
   sudo apt update
   sudo apt install python3 python3-venv python3-pip
   python3 --version
   ```

   </div>


   برای **<span dir="ltr" style="unicode-bidi: isolate;">Fedora</span>** در صورت نیاز:


   <div dir="ltr" style="direction: ltr; text-align: left;">

   ```bash
   sudo dnf install python3 python3-pip
   python3 --version
   ```

   </div>


2. از <span dir="ltr" style="unicode-bidi: isolate;">[VS Code download](https://code.visualstudio.com/download)</span> بستهٔ مناسب سیستم را بگیرید؛ روی **<span dir="ltr" style="unicode-bidi: isolate;">Ubuntu/Debian</span>** بستهٔ <span dir="ltr" style="unicode-bidi: isolate;">`.deb`</span> و روی **<span dir="ltr" style="unicode-bidi: isolate;">Fedora/RHEL</span>** بستهٔ <span dir="ltr" style="unicode-bidi: isolate;">`.rpm`</span> مناسب است. نمونهٔ نصب پس از جایگزینی نام فایل:


   <div dir="ltr" style="direction: ltr; text-align: left;">

   ```bash
   sudo apt install ./NAME_OF_FILE.deb
   # یا در Fedora:
   sudo dnf install ./NAME_OF_FILE.rpm
   ```

   </div>


3. VS Code را باز کنید و افزونه‌های **<span dir="ltr" style="unicode-bidi: isolate;">Python</span>** و **<span dir="ltr" style="unicode-bidi: isolate;">Jupyter</span>** از Microsoft را نصب کنید.

**مهم:** فرمان‌های <span dir="ltr" style="unicode-bidi: isolate;">`sudo ...`</span> مخصوص **ترمینال سیستم‌عامل** هستند، نه سلول Python. <span dir="ltr" style="unicode-bidi: isolate;">`NAME_OF_FILE`</span> یک نمونه است؛ باید با نام فایل واقعی جایگزین شود. در لینوکس بهتر است بسته‌های Python مربوط به **پروژه** را در <span dir="ltr" style="unicode-bidi: isolate;">`.venv`</span> نصب کنید، نه در Python سراسری سیستم.

راهنما: <span dir="ltr" style="unicode-bidi: isolate;">[VS Code Setup](https://code.visualstudio.com/docs/getstarted/overview)</span>

</div>

## ۷. محیط Python (Interpreter) و محیط مجازی (<span dir="ltr" style="unicode-bidi: isolate;">`.venv`</span>)

چرا ممکن است «Python نصب است» ولی VS Code خطا دهد؟ چون شاید VS Code یک **نسخه/محیط دیگر** را برای اجرا انتخاب کرده باشد.

**حداقل لازم:** Command Palette ← <span dir="ltr" style="unicode-bidi: isolate;">`Python: Select Interpreter`</span> ← محیط Python 3 مناسب را انتخاب کنید.

**روش پیشنهادی برای این درس** (یک‌بار برای پوشهٔ <span dir="ltr" style="unicode-bidi: isolate;">`Python101`</span>):

1. پوشهٔ پروژه را باز کنید.
2. Command Palette ← <span dir="ltr" style="unicode-bidi: isolate;">`Python: Create Environment`</span> ← گزینهٔ **<span dir="ltr" style="unicode-bidi: isolate;">Venv</span>** ← Python 3 نصب‌شده.
3. صبر کنید پوشه‌ای به نام <span dir="ltr" style="unicode-bidi: isolate;">`.venv`</span> ساخته شود و مطمئن شوید Interpreter منتخب به همین پوشه اشاره می‌کند.
4. برای اجرای <span dir="ltr" style="unicode-bidi: isolate;">`.ipynb`</span>، در گوشهٔ بالای دفتر روی **<span dir="ltr" style="unicode-bidi: isolate;">Select Kernel</span>** کلیک کنید ← **<span dir="ltr" style="unicode-bidi: isolate;">Python Environments</span>** ← <span dir="ltr" style="unicode-bidi: isolate;">`.venv`</span> همین پروژه. ممکن است VS Code نصب <span dir="ltr" style="unicode-bidi: isolate;">`ipykernel`</span> را پیشنهاد دهد؛ نصب را تأیید کنید.

**اگر پیام نصب <span dir="ltr" style="unicode-bidi: isolate;">`ipykernel`</span> آمد و خودکار پیش نرفت:** در ترمینال با Python **همان محیط انتخاب‌شده** بسته را نصب کنید:


<div dir="ltr" style="direction: ltr; text-align: left;">

```powershell
# Windows (بدون نیاز به فعال‌سازی محیط)
.\.venv\Scripts\python.exe -m pip install ipykernel
```

</div>



<div dir="ltr" style="direction: ltr; text-align: left;">

```bash
# macOS / Linux (بدون نیاز به فعال‌سازی محیط)
./.venv/bin/python -m pip install ipykernel
```

</div>


اگر محیط مجازی نساخته‌اید، مسیر <span dir="ltr" style="unicode-bidi: isolate;">`.venv`</span> وجود ندارد؛ ابتدا مرحلهٔ ۲ را انجام دهید. برای چند تمرین ابتدایی نصب پکیج‌های جانبی لازم نیست، اما برای اجرای سلول‌های Jupyter در محیط انتخاب‌شده، **<span dir="ltr" style="unicode-bidi: isolate;">`ipykernel`</span> باید آماده باشد**.

**فایدهٔ <span dir="ltr" style="unicode-bidi: isolate;">`.venv`</span>:** کتابخانه‌های پروژه از بقیهٔ پروژه‌ها جدا می‌مانند. آن را معمولاً به‌شکل دستی ویرایش نکنید یا برای دیگران ارسال نکنید؛ هر دانشجو روی رایانهٔ خودش می‌سازد.

راهنمای رسمی: <span dir="ltr" style="unicode-bidi: isolate;">[Python environments](https://code.visualstudio.com/docs/python/environments)</span> · <span dir="ltr" style="unicode-bidi: isolate;">[Jupyter kernels](https://code.visualstudio.com/docs/datascience/jupyter-kernel-management)</span>

### 🩺 بررسی سادهٔ محیطی که این دفتر **واقعاً** از آن استفاده می‌کند

این کد به‌جای حدس‌زدن، می‌گوید کدام فایل Python سلول‌ها را اجرا می‌کند. لازم نیست جزئیات <span dir="ltr" style="unicode-bidi: isolate;">`sys`</span> را در این جلسه حفظ کنید؛ در فصل‌های بعدی دربارهٔ import و ماژول‌ها خواهید آموخت.

````python
import sys
from pathlib import Path

print("Python:", sys.version.split()[0])
print("Interpreter:", sys.executable)
print("Working folder:", Path.cwd())
````

**چطور خروجی را بخوانیم؟**
- <span dir="ltr" style="unicode-bidi: isolate;">`Python:`</span> باید <span dir="ltr" style="unicode-bidi: isolate;">`3.x.y`</span> باشد.

- <span dir="ltr" style="unicode-bidi: isolate;">`Working folder:`</span> مکانی است که مسیرهای نسبی فایل‌ها از آن بررسی می‌شوند؛ با محل ذخیرهٔ notebook **همیشه لزوماً یکسان نیست**.

اگر بعداً <span dir="ltr" style="unicode-bidi: isolate;">`ModuleNotFoundError`</span> دیدید، **اول مسیر Interpreter را بررسی کنید**. معمولاً اشکال از نصب بسته در یک محیط و اجرای برنامه در محیط دیگر است.

برای فهمیدنِ پیام خطا، **آخرین خطِ Traceback** معمولاً نقطهٔ شروع خوبی است؛ سپس شمارهٔ خطِ اشاره‌شده را بررسی کنید.

**آزمایش امن:** در مثال زیر فقط یک متنِ عمداً اشتباه را بررسی می‌کنیم، بدون اینکه اجرای کل notebook متوقف شود. لازم نیست جزئیات <span dir="ltr" style="unicode-bidi: isolate;">`try/except`</span> را اکنون بدانید.

</div>


````python
import ast

example_with_error = 'print("Hello"'  # Missing closing parenthesis.
try:
    ast.parse(example_with_error)
except SyntaxError as problem:
    print("Detected:", problem.__class__.__name__)
    print("Reason:", problem.msg)
````

### 🪲 دیباگ‌کردن (Debugging) در VS Code

برای فایل <span dir="ltr" style="unicode-bidi: isolate;">`.py`</span>، روش پایه:
