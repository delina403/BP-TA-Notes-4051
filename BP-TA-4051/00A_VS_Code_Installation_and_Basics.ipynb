## 2. Install on Windows (Command Prompt)

1. Press the **Windows** key, type `cmd`, and open **Command Prompt**.
2. Check that Windows Package Manager is available:

   ```bat
   winget --version
   ```

3. Install Python and VS Code (one command at a time):

   ```bat
   winget install --id Python.Python.3.14 -e
   winget install --id Microsoft.VisualStudioCode -e
   ```

4. Close and reopen Command Prompt. Check the installation:

   ```bat
   python --version
   ```

If that fails, try `py --version`. You should see **Python 3.x**. In VS Code, open **Extensions** and install **Python** by **Microsoft**.

**If `winget` does not work:** Download [Python](https://www.python.org/downloads/) and [VS Code](https://code.visualstudio.com/download) from their official sites. If `python` opens the Microsoft Store, try `py` and see [Python on Windows](https://docs.python.org/3/using/windows.html).


### Install on macOS

1. Install [Python 3](https://www.python.org/downloads/) and [VS Code](https://code.visualstudio.com/download) (move the VS Code app to **Applications**).
2. Open the **Terminal** app and check:

   ```bash
   python3 --version
   ```

3. In VS Code → **Extensions**, install **Python** by **Microsoft**.

On macOS, use `python3` unless your selected environment provides `python`.


### Install on Linux

**Ubuntu / Debian:**

```bash
sudo apt update
sudo apt install python3 python3-venv python3-pip
python3 --version
```

**Fedora:**

```bash
sudo dnf install python3 python3-pip
python3 --version
```

Download [VS Code](https://code.visualstudio.com/download) for your system. Install the downloaded `.deb` or `.rpm` package using your system's package manager. For example, after opening a terminal in the Downloads folder:

```bash
# Ubuntu / Debian: replace the name with your downloaded file
sudo apt install ./your-vscode-file.deb

# Fedora: replace the name with your downloaded file
sudo dnf install ./your-vscode-file.rpm
```

In VS Code, install the **Python** extension by **Microsoft**.

## 7. Use the terminal

Open **Terminal → New Terminal** in VS Code. It usually starts in your open project folder.

| Task | Windows Command Prompt | macOS / Linux |
|---|---|---|
| Show current folder | `cd` | `pwd` |
| List files | `dir` | `ls` |
| Enter a folder | `cd Python101` | `cd Python101` |
| Go up one folder | `cd ..` | `cd ..` |
| Run `hello.py` | `py hello.py` or `python hello.py` | `python3 hello.py` |

**Terminal command vs. Python code:** Type `python hello.py` in the terminal; type `print("Hi")` **inside** `hello.py`.

A terminal may start with `C:\...>` or `$`. If you see `>>>`, you are inside Python's interactive prompt. Type `exit()` to return to the terminal.


## 8. Select the right Python interpreter

An **interpreter** is the installed Python that runs your program.

1. Open the **Command Palette**.
2. Search for **Python: Select Interpreter**.
3. Choose your installed **Python 3**.

A **virtual environment** (`.venv`) keeps packages for one project separate from other projects. To create one:

1. Open `Python101`.
2. Command Palette → **Python: Create Environment** → **Venv** → choose Python 3.
3. Use **Python: Select Interpreter** again and select `.venv`.

You do not need to install extra packages for today's programs. Keep `.venv` on your own computer; do not copy it when sharing your code.


## 9. Output and input in `.py` files

Save the following as `double.py`:

```python
text = input("Enter a whole number: ")
number = int(text)

print(number * 2)
```

Run the file in VS Code. Example:

```text
Enter a whole number: 21
42
```

- `input()` returns text (`str`).
- `int()` converts number text such as `"21"` to an integer.
- `print()` shows output. Writing `4 + 6` alone in a `.py` file does **not** print `10`; `print(4 + 6)` does.

If the user types `hello` instead of a number, `int()` raises `ValueError`.


## 10. Read errors before asking for help

| Error | What to check |
|---|---|
| `SyntaxError` | Missing quote, bracket, or colon. |
| `NameError` | Variable spelling or use before assignment. |
| `TypeError` | Wrong data types for an operation. |
| `ValueError` | Invalid input for a conversion, such as `int("hi")`. |
| `IndentationError` | Spaces at the start of the line. |
| `ModuleNotFoundError` | Missing package or wrong interpreter. |
| `FileNotFoundError` | Wrong file name or folder. |

In an error report (**traceback**), read the **last line** first, then look at the file and line number.


## 11. Debug a `.py` file

A **debugger** pauses a program so you can inspect its values.

1. Open a `.py` file in VS Code.
2. Click beside a line number to set a **breakpoint** (or press `F9`).
3. Press **F5** and choose **Python File** if prompted.
4. Inspect **Variables** or hover over a variable.
5. Use **F10** to step over a line, **F5** to continue.

Try setting a breakpoint on `print(...)` in `double.py`. See how `text` and `number` change.

## 14. Common VS Code problems

| Problem | First step |
|---|---|
| **Run** button is missing or Python will not start | Save as `.py`, install Microsoft's Python extension, select interpreter. |
| `python` command is not found | Open a fresh terminal; try `py` (Windows) or `python3` (macOS/Linux). |
| File will not run / file not found | Open the right folder; check the file name and terminal's current folder. |
| Program seems stuck | It may be waiting for `input()`; type an answer and press Enter. |
| `ModuleNotFoundError` | Check **Python: Select Interpreter** and whether the package is installed in that environment. |
| `.venv` activation is blocked in PowerShell | Select `.venv` in VS Code; you can also run `./.venv/Scripts/python.exe` in PowerShell without activating it. Do not change security settings for this lesson. |

When asking a TA for help, show the **exact error**, your **file name**, and the command you tried. Remove personal details first.
