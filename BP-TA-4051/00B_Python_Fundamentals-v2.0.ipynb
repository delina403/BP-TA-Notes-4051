## 1 · What is Python?
- **Python:** high-level, general-purpose programming language.
- **Program:** instructions saved in a `.py` file.
- **Run:** use a Python implementation (commonly **CPython**).
- Uses: mathematics, algorithms, data, automation.

````python
# A comment: ignored by Python.
2 + 3 * 4           # No output in a .py script.
print(2 + 3 * 4)    # Displays 14.
````

### Compiler ≠ interpreter
| Compiler | Interpreter / runtime |
|---|---|
| **Translates** code to another form. | **Executes** instructions. |
| May produce machine code or bytecode. | May execute bytecode / other representations. |

**Not mutually exclusive:** a language can use both.

### How CPython runs code
`.py source` → **parse & compile** → **bytecode** → **Python virtual machine** → output

- Calling Python “interpreted” does **not** mean “never compiled.”
- **CPython** is one implementation; implementations can work differently.
- `.pyc` caches may be created for imported modules.

### Editor / IDE / terminal / REPL
| Tool | Role |
|---|---|
| **Editor** (VS Code) | Write and save source files. |
| **IDE** | Editor plus coding / running / debugging tools. |
| **Terminal + shell** | Enter commands (`bash`, PowerShell, ...). |
| **Python interpreter** | Runs Python programs. |
| **REPL** (`>>>`) | Executes Python interactively. |

**A terminal is not itself the interpreter.**

### Where do commands go?
**Terminal:**
```console

```
Or `python3 demo.py` / `py demo.py` (depends on system).

**Inside `demo.py`:**
```python

```

**REPL:** `>>>` is a *prompt*, not code to copy. In `.py`, use `print` to display output.

## 2 · Variables
- A variable is a **name bound to a value**.
- `=` **assigns**; `==` **compares**.
- Evaluate **right-hand side first**, then assign the name.
- Names have no permanent declared type.

````python
x = 4
print(x)
x = x + 3
print(x)
x += 2
print(x)
````

### Assignment and rebinding
- `x = x + 1` needs an existing `x` (otherwise `NameError`).
- `y = x` binds `y` to the **current** value; changing `x` later does not change `y` here.
- Values have types; the **same name can be reused** for another type.

````python
x = 5
y = x
x = 99
print(x, y)       # 99 5
x = 2.5
print(type(x))   # <class 'float'>
````

### Multiple and augmented assignment
- `a, b = 1, 2` assigns both names.
- `a, b = b, a` **swaps** values safely.
- `a += 3` is shorthand for updating the numeric value of `a`.
- Assignment does **not** create a mathematical equation.

````python
a, b = 3, 8
print(a, b)
a, b = b, a
print(a, b)
a *= 2
print(a)
````

### Valid names
- Start with a **letter or `_`**; then letters, digits, `_`.
- **No** spaces, hyphens, leading digit, or Python **keywords** (`if`, `for`, ...).
- **Case-sensitive**: `value` ≠ `Value`. Unicode letters are legal; prefer ASCII names.
- Style: `snake_case`, descriptive names.

````python
student_count = 25
_value2 = 10
Value = 8
value = 3
print(student_count, _value2, Value, value)
# 2value = 9       # SyntaxError
# first-name = 10  # Not a variable name
# for = 3          # SyntaxError: keyword
````

### Common naming traps
- `sum = 5` is legal but hides Python's built-in `sum`.
- Likewise, don't use `print`, `input`, `int`, `float` as variable names.
- `NameError`: name not defined. `SyntaxError`: invalid code layout / keyword use.
- **Undefined** and **zero** are different.

````python
counter = 0
counter += 1
print(counter)
# print(unknown_name)   # NameError
# 4cats = 2             # SyntaxError
# counter = counter + unknown_name  # NameError
````

## 3 · Data types (today)
| Type | Example | Meaning |
|---|---|---|
| `int` | `-5`, `0`, `120` | Exact integer |
| `float` | `3.5`, `1e-3` | Approximate real number |
| `bool` | `True`, `False` | Logical truth value |
| `complex` | `2 + 3j` | Complex number |

`None` means **no value** (`NoneType`); not the number zero. **Text handling comes later.**

````python
print(type(12), type(12.0))
print(type(True), type(2 + 3j))
print(type(None))
print(3 == 3.0)              # Equal values, different types.
print(type(3) == type(3.0))
````

### Limits: what is the maximum?
| Type | Range / limits |
|---|---|
| **`int`** | **No fixed maximum or minimum**; exact, memory-limited. |
| **`float`** | Typical binary64: finite range **±1.7976931348623157e308**; ~15–17 significant decimal digits. |
| **`bool`** | Exactly **`False`** and **`True`** (numeric equivalents 0, 1). |
| **`complex`** | Real/imaginary parts follow float limits; **no meaningful ordered maximum**. |
| **`NoneType`** | Only `None`; no numeric maximum. |

**Near zero (`float`):** smallest positive *normal* ≈ `2.2250738585072014e-308`; smallest positive *subnormal* ≈ `5e-324`.

````python
import sys
import math
print(10**100 > sys.maxsize)     # sys.maxsize is NOT max int.
print(sys.float_info.max)        # Largest finite float.
print(sys.float_info.min)        # Smallest positive NORMAL float.
print(math.ulp(0.0))            # Smallest positive SUBNORMAL float.
print(int(False), int(True))
````

### Number notation
- Integer: `0`, `-7`, `1_000_000`; bases: `0b1010` (binary), `0o12` (octal), `0xA` (hex).
- Decimal `float`: `3.0`, `.5`, `1e-3` (= 0.001); scientific `e` means ×10 to a power.
- `3` ≠ `3.0` **as types**; both represent the same numerical value.
- `02` is invalid Python syntax; `0o2` is valid octal.

````python
print(0b1010, 0o12, 0xA)  # All are 10.
print(1_000_000, .5, 1e-3)
print(3 == 3.0)
# print(02)                # SyntaxError
````

### Arithmetic operators
| Symbol | Meaning | Watch |
|---|---|---|
| `+`, `-`, `*` | Add, subtract, multiply | `int` arithmetic stays exact. |
| `/` | True division | **`float` result**, even `8 / 4`. |
| `//` | Floor division | Rounds **down**, toward −∞. |
| `%` | Remainder | Linked to `//`. |
| `**` | Exponentiation | `2**3` = 8; `^` is **XOR**, not power. |

`0**0` is `1` in Python. Division by zero is an error.

````python
a, b = 7, 2
print(a + b, a - b, a * b)
print(a / b, a // b, a % b)
print(a ** b, 2 ** 3, 2 ^ 3)
print(8 / 4, 8 // 4, 8.0 // 4)
print(0 ** 0)
````

### `//` and `%` with negative numbers
- `-7 // 2` → `-4`, **not** `-3`.
- `-7 % 2` → `1`; `7 % -2` → `-1`.
- For **integers**, `a == b * (a // b) + (a % b)` when `b != 0`.
- Python remainder has the **divisor's sign** (or is zero).

````python
print(-7 / 2, -7 // 2, -7 % 2)
print(7 / -2, 7 // -2, 7 % -2)
print(-7 == 2 * (-7 // 2) + (-7 % 2))
print(divmod(-7, 2))   # (quotient, remainder)
````

### Division edge cases
- `10 / 0`, `10 // 0`, `10 % 0` → `ZeroDivisionError`.
- `0**-1` → `ZeroDivisionError`.
- `int + float` usually produces `float`; `//` on floats produces `float`.
- `2**-3` → `0.125` (**float**).

````python
print(2 + 3.0, type(2 + 3.0))
print(7.0 // 2, type(7.0 // 2))
print(2 ** -3, type(2 ** -3))
# print(1 / 0)     # ZeroDivisionError
# print(1 // 0)    # ZeroDivisionError
# print(0 ** -1)   # ZeroDivisionError
````

### Order of operations (highest → lowest)
| Priority | Operators | Rule |
|---|---|---|
| 1 | `( ... )` | Group first. |
| 2 | `**` | Powers associate **right → left**. |
| 3 | unary `+x`, `-x` | Special interaction with `**`. |
| 4 | `*`, `/`, `//`, `%` | Left → right. |
| 5 | `+`, `-` | Left → right. |
| 6 | `<`, `<=`, `>`, `>=`, `==`, `!=` | Comparisons / chains. |
| 7–9 | `not` → `and` → `or` | Decreasing priority. |

**`-2**2` = `-4`; `(-2)**2` = `4`; `2**-2` = `0.25`. Use parentheses when unclear.

````python
print(2 + 3 * 4, (2 + 3) * 4)
print(2 ** 3 ** 2, (2 ** 3) ** 2)
print(-2 ** 2, (-2) ** 2, 2 ** -2)
print(20 / 5 * 2, 20 / (5 * 2))
print(not 1 == 1, True or False and False)
````

### Comparisons
- `==`, `!=`, `<`, `<=`, `>`, `>=` → logical results.
- `0 <= x <= 10` is a **chained comparison**.
- `=` assigns; `==` compares. `is` checks **identity**, not value equality.
- For absence of value: `x is None` / `x is not None`.

````python
x = 7
print(0 <= x <= 10, x == 7, x != 7)
print(2 < 3 < 4, 2 < 3 > 4)
missing = None
print(missing is None, missing == 0)
print(3 == 3.0)
````

### Boolean logic
- Values: `True`, `False` (capital letters).
- `not` negates; `and` requires both; `or` requires either.
- **Short-circuit:** `and` may skip the second operand when first is false; `or` may skip it when first is true.
- `and` / `or` can return an **operand**, not necessarily a `bool`.

````python
print(not True, True and False, True or False)
print(bool(0), bool(3), bool(-3))
zero = 0
print(zero != 0 and (1 / zero > 2))  # Safe: skipped.
print(True or (1 / zero > 2))        # Safe: skipped.
print(5 and 0, 5 or 0)               # Returns operands.
````

### `bool` is a numeric subtype
- `False` behaves numerically like `0`; `True` like `1`.
- `bool(0)` → `False`; nonzero numbers → `True`.
- Prefer Boolean expressions for conditions, not arithmetic on `True`/`False`.
- `bool(input(...))` is **not** a reliable yes/no parser.

````python
print(True == 1, False == 0)
print(True + 2, False * 10)
print(bool(0), bool(0.0), bool(2 + 0j))
````

### Complex numbers
- `z = 3 + 4j`; `j` is the imaginary unit (`j² = -1`).
- Two `float`-based components: `z.real`, `z.imag`.
- `abs(z)` is magnitude; `z.conjugate()` changes the sign of the imaginary part.
- `<`, `>` are **not defined** for complex values (`TypeError`).

````python
z = 3 + 4j
print(type(z), z.real, z.imag)
print(abs(z), z.conjugate())
print((1j) ** 2)  # (-1+0j)
# print(z > 1)   # TypeError
````

### Type conversion
- `float(7)` → `7.0`; `int(3.9)` → `3`.
- `int(-3.9)` → `-3`: **truncates toward zero**, not floor.
- `complex(2, 3)` → `(2+3j)`; `bool(0)` → `False`.
- Invalid numeric text for `int(...)` / `float(...)` → `ValueError`.

````python
print(float(7), int(3.9), int(-3.9))
print(complex(2, 3), bool(0))
print(type(3 + 2.0), type(4 / 2))
# print(int('3.5'))  # ValueError; do not run.
````

### Rounding ≠ conversion
- `int(-3.9)` = `-3`; `math.floor(-3.9)` = `-4`; `math.ceil(-3.9)` = `-3`.
- `round(2.5)` = `2`, `round(3.5)` = `4` (**ties to even**).
- `round(x, 2)` still returns a number; it does **not** guarantee two displayed decimal places.
- Approximate stored floats affect some rounding examples.

````python
import math
print(int(-3.9), math.floor(-3.9), math.ceil(-3.9))
print(round(2.5), round(3.5))
print(round(2.675, 2))
print(round(2.0, 2))
````

### Floating-point accuracy
- `float` stores **binary approximations**; most decimal fractions are not exact.
- `0.1 + 0.2 != 0.3` on typical CPython.
- Use `math.isclose(a, b)` for *approximate* numeric comparisons (choose tolerances for your problem).
- `==` is fine when **exact equality** is intended.

````python
import math
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
print(math.isclose(0.1 + 0.2, 0.3))
print(1.0 == 1.0 + 1e-16)  # Precision is finite.
````

### Float extremes & special values
- Largest finite ≈ `1.8e308`; overflow may yield **`inf`** (some operations raise an error).
- Tiny magnitudes may **underflow** to `0.0`.
- **`nan`** means “not a number”: it does **not** equal itself.
- `math.isfinite(...)` tests whether a float is finite.

````python
import math
print(1e309, -1e309)           # inf, -inf
print(1e-400)                 # 0.0
print(math.nan == math.nan)   # False
print(math.isfinite(1e309), math.isfinite(3.0))
````

### `int` is exact — with practical limits
- `10**100` is an ordinary Python `int`; **`sys.maxsize` is not its maximum**.
- Integer size is bounded by available **memory/time**.
- Very large integers converted to `float` may raise `OverflowError`.
- Python 3.11+ may limit **very long decimal input/output conversions**; that is *not* an `int` value limit.

````python
import sys
big = 10 ** 100
print(big > sys.maxsize)
print(big // 10 ** 99)   # Exact: 10.
print(10 ** 30 + 1 - 10 ** 30)
# print(float(10 ** 1000))  # OverflowError on typical CPython.
````

## 4 · Input / output
**`input()` → convert → compute → `print()`**

- In a **`.py` program**, `print(...)` displays values.
- `input(...)` pauses until a line is entered; its result is **text, not a number**.
- For numeric calculations: `int(input(...))` or `float(input(...))`.

### Output: `print(...)`
- `print(a, b)` separates arguments with a **space**; ends with a **newline**.
- `sep=` changes the separator; `end=` changes the ending.
- `print()` alone outputs a blank line.
- Quoted labels are used for display; **text operations are not covered today**.

````python
result = 7 + 3
print(result)
print("Result:", result)
print(1, 2, 3)
print(1, 2, sep="-", end="; ")
print(3)
print()
````

### Input: choose a number type
- `int(input(...))` for **whole numbers**, e.g. `12`, `-7`.
- `float(input(...))` for **decimals**, e.g. `3.5`, `1e-3`.
- `int` rejects decimal input like `3.5` → `ValueError`.
- `float` normally uses a **dot** (`3.5`), not a comma (`3,5`).

````python
# LIVE in terminal or notebook — type a whole number.
age = int(input("Age: "))
print("Next year:", age + 1)
````

````python
# LIVE — type two decimal numbers.
a = float(input("a: "))
b = float(input("b: "))
print("Sum:", a + b)
print("Mean:", (a + b) / 2)
````

### Numerical output: one pitfall
- `print(2 / 4)` → `0.5`; `print(2, 4)` → `2 4`.
- `round(x, 2)` changes the returned numeric value, but **does not force two trailing zeros** on screen.
- Display formatting is a separate topic; use `round` only when rounding is intended.

````python
x = 2.0
print(x, round(x, 2))    # 2.0 2.0, not 2.00.
print(2 / 4, 2, 4)
print("Area:", round(3.5 * 2, 2))
````

### Live program 1 · Time decomposition
