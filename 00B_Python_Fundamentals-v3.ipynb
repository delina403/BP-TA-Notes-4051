## 1. Introduction to Programming

### 1.1. Algorithms and programs
**Programming** is the process of expressing instructions in a language that a computer can execute.

An **algorithm** is a well-defined sequence of steps for solving a problem. A **program** implements instructions in a programming language.

For example, an algorithm for calculating the arithmetic mean of two numbers:
1. Obtain two numbers.
2. Add them.
3. Divide their sum by two.
4. Display the result.

An algorithm describes the solution; a program implements it.

### 1.2. Python
Python is a high-level, general-purpose programming language designed to be readable and expressive. It is widely used in mathematics, scientific computing, data analysis, artificial intelligence, automation and software development.

A **high-level language** lets programmers express instructions without directly managing most hardware-level operations.

### 1.3. Program execution
Instructions written by a programmer are called **source code**.

| Term | Description |
|---|---|
| Compiler | Translates code into another representation. |
| Interpreter | Executes instructions expressed in a programming language or an intermediate representation. |

Compilation and interpretation are not mutually exclusive; an implementation may use both. In a simple sequential program, instructions execute from top to bottom. The built-in `print()` function displays values.

````python
print(10)
print(20)
print(30)
````

### 1.4. Input–Processing–Output
Many programs can be described using three stages:

```text
┌─────────┐     ┌────────────┐     ┌─────────┐
│  Input  │ ──► │ Processing │ ──► │ Output  │
└─────────┘     └────────────┘     └─────────┘
```

For a program calculating the area of a rectangle:

| Stage | Description |
|---|---|
| Input | Width and height |
| Processing | Multiply width by height |
| Output | Area |

## 2. Basic Python Syntax
**Syntax** refers to the grammatical rules that determine how valid code must be written.

### 2.1. Literals, expressions and statements
A **literal** is a notation representing a value directly in the source code, such as `10` or `3.14`.

An **expression** is a piece of code that can be evaluated to produce a value, such as `2 + 3`.

A **statement** is an executable instruction, such as `x = 2 + 3`. Here, `2 + 3` is an expression, while the entire line is an assignment statement. A standalone expression can also be an expression statement.

````python
10               # Literal
3.14             # Literal
x = 2 + 3        # Assignment statement
2 + 3            # Expression statement (displayed by the notebook)
````

### 2.2. Comments
Comments provide explanations for people reading the code. A comment begins with `#` and continues to the end of the line (outside a string). Python ignores comments during execution; they can also be used to temporarily disable instructions.

````python
# Display numbers
print(10)
# print(20)
print(30)  # Display the final number
````

### 2.3. Case sensitivity
Python is **case-sensitive**: uppercase and lowercase letters are treated differently. `print` is a built-in name; `Print` is not.

````python
print(10)  # Correct
# Print(10)  # NameError if this line is uncommented
````

### 2.4. Whitespace and indentation
Spaces and blank lines improve readability. These two instructions have the same result, but the second follows better coding conventions.

````python
print(2+3)
print(2 + 3)
````

**Indentation** is whitespace at the beginning of a line. Python uses indentation to define blocks of code. For now, do not unnecessarily indent top-level instructions.

````python
print(10)
print(20)
````

Incorrect indentation (uncomment the second line to demonstrate):

## 3. Using Built-in Functions
A **function** is a callable unit of code designed to perform a task. Python provides ready-to-use **built-in functions** without requiring imports.

General form of a function call:

```text
function_name(arguments)
```

An **argument** is a value or expression supplied to a function.

### 3.1. `print()` and `type()`
`print()` displays values, while `type()` returns the type of a value. Multiple arguments are separated by commas. Expressions supplied as arguments are evaluated before the function is called.

````python
print(10)
print(10, 20, 30)
print(2 + 3)

print(type(10))
print(type(3.14))
````

For now, focus on the type names `int` and `float` in the output.

### 3.2. Nested function calls
A function call can appear inside another call. Python evaluates the inner call first, then passes its result to the outer function.

```text
       10
        │
        ▼
     type()
        │
        ▼
   <class 'int'>
        │
        ▼
     print()
```

````python
print(type(10))
````

The result produced by a function call is called its **return value**. Returning a value and displaying a value are different operations.

## 4. Variables and Assignment

### 4.1. Variables
A variable is a **name bound to a value**. Variables allow us to access values without repeatedly writing them directly.

```text
Name          Value
 x ─────────►   5
```

````python
x = 5
print(x)
````

### 4.2. Assignment
The symbol `=` is the assignment operator.

```text
variable = expression
```

Python first evaluates the right-hand side, then binds the name on the left-hand side to the resulting value. Defined variables can be used in other expressions.

````python
x = 2 + 3
y = 7

result = x + y

print(result)
````

### 4.3. Reassignment
A variable can be assigned a new value.

```text
Before reassignment:   x ─────► 5

After reassignment:    x ─────► 12
```

````python
x = 5
print(x)

x = 12
print(x)
````

### 4.4. Assignment between variables

````python
x = 5
y = x

x = 10

print(x, y)
````

Initially, `x` and `y` refer to the same value:

```text
x ───┐
     ├────► 5
y ───┘
```

After reassigning `x`:

```text
x ────────► 10

y ────────► 5
```

Assignment does not establish a permanent mathematical relationship.

````python
x = 3
y = 2 * x

x = 10

print(y)
````

Python does not automatically recalculate `y` when `x` changes. Unlike a mathematical equation, assignment is an instruction executed at a particular point in time.

### 4.5. Tracing variables
Manually tracking variable values throughout execution is called **manual tracing** or a **dry run**. It is useful for understanding and debugging programs.

````python
a = 3
b = 7
c = a + b
a = 10
b = c

print(a, b, c)
````

| Instruction executed | `a` | `b` | `c` |
|---|---:|---:|---:|

The dash indicates a name not yet assigned in this example (assuming a fresh namespace).

## 5. Fundamental Data Types
A **data type** determines what kind of value we are working with and which operations can be performed on it.

| Type | Description | Examples |
|---|---|---|
| `int` | Integer | `12`, `-5`, `0` |
| `float` | Floating-point number | `3.14`, `-2.5` |

| `str` | String (text) | `"Hello"`, `"12"` |

### 5.1. Bits and bytes
A **bit** is a binary digit (`0` or `1`). A **byte** consists of eight bits.

```text
1 byte = 8 bits

┌───┬───┬───┬───┬───┬───┬───┬───┐
│ 0 │ 1 │ 0 │ 1 │ 1 │ 0 │ 0 │ 1 │
└───┴───┴───┴───┴───┴───┴───┴───┘
```

Binary digits correspond to powers of 2, just as decimal digits correspond to powers of 10:

$$(1101)_2 = 1(2^3)+1(2^2)+0(2^1)+1(2^0)=13.$$

Using $n$ bits, there are $2^n$ distinct patterns, or nonnegative integers from $0$ to $2^n-1$ when interpreted as unsigned binary numbers.

| Bits | Bytes (if whole) | Possible patterns | Maximum unsigned integer |
|---:|---:|---:|---:|
| 1 | — | 2 | 1 |
| 2 | — | 4 | 3 |
| 4 | — | 16 | 15 |
| 8 | 1 | 256 | 255 |
| 16 | 2 | 65,536 | 65,535 |
| 32 | 4 | 4,294,967,296 | 4,294,967,295 |

These are general binary ranges, not limits on Python's `int` type.

### 5.2. Integers — `int`
Integers represent positive and negative whole numbers, including zero. Python integers have **arbitrary precision**: their size can grow as needed, subject to available resources. Underscores improve the readability of large numerical literals without changing their values.

````python
a = 15
b = -8
c = 0

print(type(a), type(b), type(c))

number = 1234567890123456789012345678901234567890
population = 1_000_000

print(number)
print(population)
````

### 5.6. Data type capacity

| Type | Representation | Maximum digits or values |
|---|---|---|
| `int` | Variable number of bits / bytes | No fixed numerical digit limit |

Significant digits count meaningful digits, not just those after the decimal point. These figures describe value representation, not total Python object size.

## 6. Arithmetic Operators
An **operator** performs an operation on one or more values, called **operands**.

```text
      7          +          3
      ↑          ↑          ↑
   Operand    Operator   Operand
```

### 6.1. Arithmetic operators
| Operator | Operation | Example | Result |
|---|---|---|---:|
| `+` | Addition | `7 + 2` | 9 |
| `-` | Subtraction | `7 - 2` | 5 |
| `*` | Multiplication | `7 * 2` | 14 |
| `/` | True division | `7 / 2` | 3.5 |
| `//` | Floor division | `7 // 2` | 3 |
| `%` | Modulo | `7 % 2` | 1 |
| `**` | Exponentiation | `7 ** 2` | 49 |

````python
a = 7
b = 2

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)
````

A **unary operator** acts on one operand, while a **binary operator** acts on two.

````python
print(-5)     # Unary negation
print(+5)     # Unary plus
print(7 - 2)  # Binary subtraction
````

### 6.2. True division — `/`
For integer and floating-point operands, true division produces a floating-point result, even when the mathematical quotient is an integer.

````python
print(7 / 2)
print(8 / 4)
print(type(8 / 4))
````

### 6.3. Floor division — `//`
For integers $a,b$, where $b\neq0$:

$$a\mathbin{//}b=\left\lfloor\frac{a}{b}\right\rfloor.$$

Floor division rounds the quotient downward toward negative infinity, not toward zero.

````python
print(7 // 2)
print(10 // 3)
print(-7 // 2)
````

#### Negative floor division
```text
 -5         -4    -3.5    -3         -2
  │──────────●─────×───────│──────────│
             ◄─────┘
              Floor
```

$$\lfloor-3.5\rfloor=-4.$$

Floor division does not simply remove the fractional part.

### 6.4. Modulo — `%`
Modulo produces the remainder associated with floor division. For integers $a,b$, where $b\neq0$:

$$a=bq+r,\qquad q=a\mathbin{//}b,\qquad r=a\bmod b.$$

For a positive divisor $b$, $0\leq r<b$. For example, $17=5(3)+2$.

````python
a = 17
b = 5

q = a // b
r = a % b

print(q, r)
print(b * q + r)
````

#### Negative operands
In Python, a nonzero integer remainder has the same sign as the divisor.

| `a` | `b` | `a // b` | `a % b` |
|---:|---:|---:|---:|
| 7 | 2 | 3 | 1 |
| -7 | 2 | -4 | 1 |
| 7 | -2 | -4 | -1 |
| -7 | -2 | 3 | -1 |

For example, $-7=2(-4)+1$.

````python
print(-7 // 2, -7 % 2)
print(7 // -2, 7 % -2)
````

### 6.5. Applications of floor division and modulo
**Digit extraction** for nonnegative integers:

````python
number = 5837

print(number % 10)   # Last digit
print(number // 10)  # Remove the last digit
````

**Divisibility:** a zero remainder indicates divisibility.

````python
number = 24

print(number % 2)
print(number % 3)
print(number % 5)
````

### 6.6. Exponentiation — `**`
Negative exponents represent reciprocals, and fractional powers can represent roots of nonnegative numbers.

$$2^{-3}=\frac{1}{2^3},\qquad 25^{1/2}=\sqrt{25},\qquad16^{1/4}=\sqrt[4]{16}.$$

````python
print(2 ** 3)
print(2 ** -3)
print(25 ** 0.5)
print(16 ** 0.25)
````

**Common mistake:** `^` does not represent exponentiation in Python. Use `**`.

### 6.7. Arithmetic result types
The result type depends on both the operands and the operation.

| Expression | Result | Type |
|---|---:|---|
| `3 + 2` | 5 | `int` |
| `3 + 2.0` | 5.0 | `float` |
| `8 / 4` | 2.0 | `float` |

| `8.0 // 4` | 2.0 | `float` |
| `2 ** -3` | 0.125 | `float` |

````python
print(type(3 + 2.0))
print(type(8 / 4))
print(type(8.0 // 4))
````

## 7. Operator Precedence and Associativity

### 7.1. Precedence
Precedence determines how expressions containing different operators are grouped.

From highest to lowest:

| Priority | Operator or grouping |
|---:|---|
| 1 | Parentheses `(...)` (explicit grouping) |
| 2 | Exponentiation `**` |
| 3 | Unary `+x`, `-x` |
| 4 | `*`, `/`, `//`, `%` |
| 5 | `+`, `-` |

Exponentiation has a special interaction with unary operators, explained below.

````python
print(2 + 3 * 4)
print((2 + 3) * 4)
````

### 7.2. Associativity
Associativity determines grouping when consecutive operators have equal precedence. Most arithmetic binary operators associate from left to right.

````python
print(20 / 5 * 2)
print((20 / 5) * 2)
print(20 / (5 * 2))
````

Exponentiation associates from right to left.

````python
print(2 ** 3 ** 2)
print(2 ** (3 ** 2))
print((2 ** 3) ** 2)
````

### 7.3. Unary minus and exponentiation
Exponentiation binds more tightly than unary minus on its left, but a unary minus can directly appear in the exponent.

$$-(2^2)=-4,\qquad (-2)^2=4,\qquad 2^{-2}=0.25.$$

````python
print(-2 ** 2)
print((-2) ** 2)
print(2 ** -2)
````

**Good practice:** Use parentheses whenever they make the intended grouping clearer.

## 8. Updating Variables

### 8.1. Updating using previous values
A variable can appear on both sides of an assignment. Python evaluates the right-hand side using the previous value before assigning the new value.

```text
Old value:       x = 5
Calculation:     5 + 1 = 6
New assignment:  x = 6
```

Although $x=x+1$ is contradictory as an ordinary mathematical equation, it is meaningful as a programming instruction.

````python
x = 5
x = x + 1

print(x)
````

### 8.2. Augmented assignment
Augmented assignment provides a shorter notation for updating variables. For the numerical types covered in this lesson:

| Augmented assignment | Ordinary assignment |
|---|---|
| `x += 2` | `x = x + 2` |
| `x -= 2` | `x = x - 2` |
| `x *= 2` | `x = x * 2` |
| `x /= 2` | `x = x / 2` |
| `x //= 2` | `x = x // 2` |
| `x %= 2` | `x = x % 2` |
| `x **= 2` | `x = x ** 2` |

````python
x = 10

x += 5
x *= 2
x -= 4

print(x)
````

Updating a variable can change the type of its value.

````python
x = 10
x /= 2

print(x, type(x))
````

### 8.3. Initialization
A variable must have an initial value before it can be updated. Without initialization, the update may attempt to access an undefined variable.

````python
counter = 0

counter += 1
counter += 1

print(counter)
````

### 8.4. Swapping variables
Incorrect approach:

````python
a = 5
b = 8

a = b
b = a

print(a, b)
````

The original value of `a` has been lost. Correct approach using a temporary variable:

````python
a = 5
b = 8

temp = a
a = b
b = temp

print(a, b)
````

## 9. Type Conversion
Type conversion is the process of obtaining a value of one type from a value of another type.

| Operation | Purpose |
|---|---|
| `int()` | Obtain an integer |
| `float()` | Obtain a floating-point number |
| `str()` | Obtain a string |

### 9.1. Numerical conversion

````python
print(float(7))
print(int(3.9))
print(int(-3.9))
````

Converting a finite floating-point number to an integer **truncates toward zero**. It does not round to the nearest integer or necessarily round downward. Large integers may lose precision when converted to floating-point numbers.

````python
print(int(-3.9))
print(-3.9 // 1)
````

### 9.2. Converting between strings and numbers

````python
a = int("25")
b = float("3.14")
c = str(25)

print(a, type(a))
print(b, type(b))
print(c, type(c))
````

Not every string represents a valid number. The following conversions produce a `ValueError` when executed.

````python
# int("hello")
# int("3.5")
# float("hello")
# int("")
````

Notice the distinction between converting a number and converting numerical text. When truncation is intended, two conversions can be performed.

````python
print(int(3.5))           # Valid
# int("3.5")              # Invalid
print(int(float("3.5")))  # Valid
````

### 9.3. Conversion and reassignment
Conversion does not automatically update the original variable. Assign the returned value to update it.

````python
x = "12"

int(x)
print(type(x))

x = int(x)
print(type(x))
````

### 9.4. Implicit numerical conversion
Python automatically handles certain operations involving different numerical types. It does not automatically treat numerical text as a number.

````python
result = 3 + 2.5
print(result, type(result))

# print("12" + 3)  # TypeError
print(int("12") + 3)
````

## 11. Floating-Point Precision
Floating-point numbers have finite precision and cannot represent every real number exactly. For example, a finite decimal representation cannot exactly represent

$$\frac{1}{3}=0.33333333\ldots$$

Similarly, some decimal fractions cannot be represented exactly using a finite number of binary digits.

### 11.1. Floating-point approximation

````python
print(1 / 3)
print(0.1 + 0.2)
print(0.5 + 0.25)
````

The values `0.1` and `0.2` are stored as binary approximations, whereas `0.5` and `0.25` can be represented exactly.

### 11.2. Precision and magnitude

````python
print(1_000_000_000_000_000_000 + 1)
print(1_000_000_000_000_000_000.0 + 1)
````

The integer preserves the added unit, but the floating-point number cannot distinguish such a small change relative to its magnitude.

### 11.3. Rounding floating-point results

````python
result = 0.1 + 0.2

print(result)
print(round(result, 2))
````

Rounding may produce a more suitable approximation, but it does not eliminate the underlying limitations of floating-point arithmetic.

**General principle:** Use integers for exact integer arithmetic and floating-point numbers for approximate real-valued quantities.

## 12. Comparison Operators
Comparison operators compare values and produce Boolean results.

| Python | Mathematics | Meaning |
|---|---|---|
| `==` | $=$ | Equal to |
| `!=` | $\neq$ | Not equal to |
| `>` | $>$ | Greater than |
| `<` | $<$ | Less than |
| `>=` | $\geq$ | Greater than or equal to |
| `<=` | $\leq$ | Less than or equal to |

````python
print(5 == 5)
print(5 != 3)
print(5 > 3)
print(5 < 3)
print(5 >= 5)
print(5 <= 4)
````

### 12.1. Assignment vs. equality
Assignment (`=`) binds a name to a value. Equality comparison (`==`) determines whether two values are equal.

````python
x = 5           # Assignment

print(x == 5)   # Equality comparison
print(x == 10)
````

### 12.2. Storing comparison results
Comparison expressions produce Boolean values that can be assigned to variables. Arithmetic operators have higher precedence than comparison operators.

````python
x = 7

is_positive = x > 0
is_even = x % 2 == 0

print(is_positive)
print(is_even)
print(type(is_even))

print(2 + 3 > 4)
print((2 + 3) > 4)
````

### 12.3. Comparing different types
Values can be numerically equal despite having different numerical types. Numerical text, however, is not the same as a number.

````python
print(5 == 5.0)
print(5 == "5")
````

### 12.4. Chained comparisons
Python supports mathematical-style chained comparisons, useful for testing whether a value belongs to an interval.

| Mathematical interval | Python |
|---|---|
| $x\in(a,b)$ | `a < x < b` |
| $x\in[a,b]$ | `a <= x <= b` |
| $x\in[a,b)$ | `a <= x < b` |
| $x\in(a,b]$ | `a < x <= b` |

````python
x = 7

print(0 < x < 10)
print(0 <= x <= 7)
print(10 < x < 20)
````

### 12.5. Comparing floating-point numbers

````python
print(0.1 + 0.2 == 0.3)
````

When comparing approximate quantities, we can examine their absolute difference.

$$|a-b|<\varepsilon$$

The positive tolerance $\varepsilon$ is chosen according to the required accuracy and scale of the problem.

````python
a = 0.1 + 0.2
b = 0.3

difference = abs(a - b)

print(difference)
print(difference < 0.000001)
````

#### Escape sequences — `\n` and `\t`

Inside a quoted string, a backslash can introduce an **escape sequence**. It represents a special character rather than ordinary text.

| Sequence | Meaning | Effect when printed |
|---|---|---|
| `\n` | Newline | Moves to the beginning of the next line |
| `\t` | Horizontal tab | Moves to the next tab position |

Each sequence represents **one character**, not two. A tab is not a fixed number of spaces; its appearance depends on the display. `print()` normally adds a newline at the end, while `\n` inside a string inserts an additional newline wherever it appears.

These sequences also work in `sep` and `end`. To print the backslash literally, write `\\` (for example, `"A\\nB"`). A notebook may show escape notation when displaying a string automatically; `print()` displays the actual text.

````python
print("First line\nSecond line")
print("Name\tScore\nMina\t18")

print(10, 20, 30, sep="\n")
print(10, 20, 30, sep="\t")
print("A", end="\t")
print("B")

print("A\\nB")  # Literal backslash and n
````

Rounding changes a numerical value; formatting controls how it is displayed. Displaying `2.00` requires output formatting, which will be studied in the strings lesson.

## Appendix — ASCII and Character Representation
*Supplementary classroom material; may be deferred to the strings lesson if time is limited.*

### A.1. How do computers represent characters?
Computers represent information using binary digits. To represent text, characters must be associated with numerical codes. A **character encoding** establishes rules for representing characters numerically.

**ASCII** stands for **American Standard Code for Information Interchange**. It assigns numerical codes to English letters, digits, punctuation marks and certain control characters. ASCII uses 7 bits, allowing $2^7=128$ distinct character codes, from 0 to 127.

### A.2. Character representation
For example, the uppercase letter `A` has ASCII code 65.

```text
Character        Decimal          Binary

   A       ──►      65      ──►   1000001
```

An ASCII code can be stored in one byte:

```text
          ASCII character: A

┌───┬───┬───┬───┬───┬───┬───┬───┐
│ 0 │ 1 │ 0 │ 0 │ 0 │ 0 │ 0 │ 1 │
└───┴───┴───┴───┴───┴───┴───┴───┘
 128  64  32  16   8   4   2   1

            64 + 1 = 65
```

### A.3. Important ASCII ranges
| Decimal codes | Characters |
|---|---|
| 0–31 | Control characters |
| 32 | Space |
| 33–47, 58–64, 91–96, 123–126 | Punctuation and symbols |
| 48–57 | Digits `0`–`9` |
| 65–90 | Uppercase letters `A`–`Z` |
| 97–122 | Lowercase letters `a`–`z` |
| 127 | Delete control character |

Selected ASCII codes:

| Character | Decimal | Binary (8-bit display) |
|---|---:|---|
| Space | 32 | `00100000` |
| `0` | 48 | `00110000` |
| `1` | 49 | `00110001` |
| `9` | 57 | `00111001` |
| `A` | 65 | `01000001` |
| `B` | 66 | `01000010` |
| `Z` | 90 | `01011010` |
| `a` | 97 | `01100001` |
| `b` | 98 | `01100010` |
| `z` | 122 | `01111010` |

Uppercase and lowercase letters have different codes.

### A.4. Obtaining character codes — `ord()`
Python's built-in `ord()` function returns the Unicode code point of a single character. For ASCII characters, this is also their ASCII code.

````python
print(ord("A"))
print(ord("a"))
print(ord("0"))
print(ord(" "))
````

`ord()` expects exactly one character when given a string.

````python
# ord("Hello")  # TypeError
````

### A.5. Obtaining characters — `chr()`
The built-in `chr()` function returns the character associated with a Unicode code point.

```text
           ord()
     "A" ────────► 65

           chr()
     "A" ◄──────── 65
```

````python
print(chr(65))
print(chr(97))
print(chr(48))
print(chr(ord("A")))
````

### A.6. Patterns in ASCII
The codes for uppercase letters are consecutive, as are those for lowercase letters and digits.

```text
Uppercase:   A   B   C   D   ...   Z
            65  66  67  68       90

Lowercase:   a   b   c   d   ...   z
            97  98  99  100     122

Digits:      0   1   2   3   ...   9
            48  49  50  51       57
```

**Example — The next letter:**

````python
character = "A"
next_character = chr(ord(character) + 1)

print(next_character)
````

**Example — Uppercase and lowercase:** Corresponding uppercase and lowercase ASCII letters differ by 32.

````python
print(ord("a") - ord("A"))

print(chr(ord("B") + 32))
print(chr(ord("b") - 32))
````

This relationship applies to the English letters in ASCII, not to all characters in other languages.

### A.7. Characters vs. numerical values
The character `"7"` and the integer `7` are different values. `int()` interprets numerical text as a number; `ord()` returns the character's numerical code.

````python
print(int("7"))
print(ord("7"))
````

### A.8. ASCII and Unicode
ASCII defines only 128 character codes. It cannot represent Persian letters, most other writing systems or emojis.

**Unicode** is a much broader standard assigning numerical code points to characters from many writing systems. Modern Python strings support Unicode, and its first 128 code points match ASCII.

````python
print("Hello")
print("سلام")
print("π")
print("∞")
````
